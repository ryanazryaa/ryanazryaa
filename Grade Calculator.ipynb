```python
komponen = int(input("Masukkan banyak komponen : "))

def input_nilai():
  nama = input("Masukkan Nama : ")
  nilai_Tugas = int(input("Masukkan Nilai Tugas : "))
  nilai_UTS = int(input("Masukkan Nilai UTS : "))
  nilai_UAS = int(input("Masukkan Nilai UAS : "))
  bobot_Tugas = int(input("Masukkan Bobot Tugas : "))
  bobot_UTS = int(input("Masukkan Bobot UTS : "))
  bobot_UAS = int(input("Masukkan Bobot UAS : "))

  if bobot_Tugas + bobot_UTS + bobot_UAS != 100:
    print("Bobot Tidak Sesuai")


  nilai_akhir = float(nilai_Tugas * (bobot_Tugas / 100)) + (nilai_UTS * (bobot_UTS / 100)) + (nilai_UAS * (bobot_UAS / 100))
  huruf = nilai_huruf(nilai_akhir)
  print(f"Nama : {nama}")
  print(f"Nilai Akhir : {nilai_akhir}")
  print(f"Nilai Huruf : {huruf}")

def nilai_huruf(nilai_akhir):
  if nilai_akhir >= 85:
    return "A"
  elif nilai_akhir >= 75:
    return "AB"
  elif nilai_akhir >= 65:
    return "B"
  elif nilai_akhir >= 60:
    return "BC"
  elif nilai_akhir >= 50:
    return "C"
  elif nilai_akhir >= 45:
    return "D"
  else:
    return "E"

for i in range(komponen):
  input_nilai()
```
